## 1. Import and Load Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_FILE = '1776311302-P3-Car Market Trends Analysis with Car Dekho Data.csv'
df = pd.read_csv(DATA_FILE)
df.head()
```

## 2. Data Quality

```python
print('Shape:', df.shape)
print('\nMissing values:\n', df.isna().sum())
print('\nDuplicate rows:', df.duplicated().sum())
```

## 3. Clean and Create Derived Columns

```python
data = df.drop_duplicates().copy()
data['Car_Age'] = 2018 - data['Year']
data['Depreciation'] = data['Present_Price'] - data['Selling_Price']
data['Resale_Percentage'] = data['Selling_Price'] / data['Present_Price'] * 100
data.describe()
```

## 4. Market Analysis

```python
print('Average selling price:', round(data['Selling_Price'].mean(), 2), 'lakh')
print('Median selling price:', round(data['Selling_Price'].median(), 2), 'lakh')
print('\nFuel type:')
print(data.groupby('Fuel_Type')['Selling_Price'].mean().sort_values(ascending=False))
print('\nSeller type:')
print(data.groupby('Seller_Type')['Selling_Price'].mean().sort_values(ascending=False))
print('\nTransmission:')
print(data.groupby('Transmission')['Selling_Price'].mean().sort_values(ascending=False))
```

## 5. Visualizations

```python
year = data.groupby('Year')['Selling_Price'].mean()
plt.figure(figsize=(8,4.5))
plt.plot(year.index, year.values, marker='o')
plt.xlabel('Manufacturing Year')
plt.ylabel('Average Selling Price (₹ lakh)')
plt.title('Average Selling Price by Manufacturing Year')
plt.grid(alpha=0.25)
plt.show()
```

```python
plt.figure(figsize=(7,4.5))
plt.scatter(data['Present_Price'], data['Selling_Price'], alpha=0.45)
plt.xlabel('Present Price (₹ lakh)')
plt.ylabel('Selling Price (₹ lakh)')
plt.title('Present Price vs Selling Price')
plt.show()
```

## 6. Interpretation

The results are descriptive findings from the supplied dataset. Group-level differences should not be interpreted as causal effects.
